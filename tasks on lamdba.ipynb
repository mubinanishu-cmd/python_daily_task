#**30 Python Lambda Function practice questions**

#**Easy Level (1–10)**

```python
# 1.Write a lambda function to return the square of a number.

def square(x):
    return x * x
print(square(5))
```

```python
# 2.Write a lambda function to return the cube of a number.

def cube(x):
    return x * x * x
print(cube(4))
```

```python
# 3. Write a lambda function to add two numbers.

def add(x):
    return x + 10
print(add(20))
```

```python
# 4. Write a lambda function to find the maximum of two numbers.

num = [15, 10]
largest = max(num,key=lambda x : x)
print("Largest:",largest)
```

```python
# 5. Write a lambda function that returns "Even" if a number is even; otherwise "Odd".

check = lambda n : "Even" if n % 2 == 0 else "Odd"
print(check(7))
print(check(8))
```

```python
# 6. Use map() with lambda to square every element of a list.

numbers = [1,2,3,4]
result = list(map(lambda x: x**2,numbers))
print(result)
```

```python
# 7.Use filter() with lambda to keep only even numbers.

num = [1,2,3,4,5,6]
result = list(filter(lambda x: x % 2==0,num))
print(result)
```

```python
# 8. Use filter() with lambda to keep numbers greater than 50.

numbers = [20,45,55,70]
result = list(filter(lambda x: x>= 50,numbers))
print(result)
```

```python
# 9. Sort a list of numbers in ascending order using lambda.

num = [9,5,1,8,2]
result = sorted(num,key=lambda x : x, reverse=False)
print(result)
```

```python
# 10 .Sort a list of numbers in descending order using lambda.

num = [9,5,1,8,2]
result = sorted(num,key = lambda x: x,reverse=True)
print(num)
```

```python

#28.Find the second highest salary using lambda.
employees = [
    ("A", 45000),
    ("B", 60000),
    ("C", 52000),
    ("D", 75000)
]

sorted_employees = sorted(employees, key=lambda x: x[1], reverse=True)

print("Second highest salary:", sorted_employees[1])
```

```python
# 29.Sort a list of dictionaries by age ascending and then salary descending.
employees = [
{"name":"Ram","age":25,"salary":50000},
{"name":"Amit","age":25,"salary":60000},
{"name":"Neha","age":23,"salary":45000},
{"name":"Raj","age":27,"salary":55000}
]
result = sorted(employees, key=lambda x: (x["age"], -x["salary"]))
print(result)
```
