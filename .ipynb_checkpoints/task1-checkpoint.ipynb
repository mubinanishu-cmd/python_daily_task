```python
#Q5. Arithmetic Operators
# Take two numbers from user and print:
# Sum, Difference, Product, Division, Floor Division, Modulus, Power
x = 10
y = 5
print( x + y)   #Sum
print( x - y)   #Difference
print( x * y)   #Product
print( x / y)   #Division
print( x ** y)  #Power
print( x // y)  #Floor Division
print( x % y)   #Modulus
```

```python
#Q6. Comparison Operators
# Take two numbers and check:
# Equal, Not Equal, Greater, Smaller, Greater or Equal, Smaller or Equal
x = 5
y = 12
print( x==y)  #Equal
print( x != y) #Not equal
print( x > y) #Greater
print( x < y)  #Smaller
print( x >= y) #Greater than
print( x <= y)   #Smaller than
```

```python
#Simple Calculator
# Create a simple calculator that takes two numbers and one operator (+,-,*,/,//,%,**)
# from the user and shows the result.

num1 = int(input("Enter the first number:"))
num2 = int(input("Enter the second number:"))
operator = input("Enter the operator(+,-,*,/,//,%,**)")

if operator == "+":
    result = num1+num2
elif operator =="-":
    result = num1-num2
elif operator =="*":
    result = num1*num2
elif operator =="/":
    result = num1/num2
elif operator =="//":
    result = num1//num2
elif operator =="%":
    result = num1%num2
elif operator =="**":
    result = num1**num2
else:
    result = "Invalid operator"
print("Result:",result)
```

```python
#Q10:Temperature Converter
# Convert temperature from Celsius to Fahrenheit and vice versa.
# Formula: F = (C * 9/5) + 32

temp = float(input("Enter the temperature: "))
choice = input("Convert to Celsius or Fehrenheit?(C/F): ")

if choice.upper() =="F":
    result = (temp * 9/5) + 32
    print("Temperature in Fehrenheit:",result)
elif choice.upper() =="C":
    result =(temp - 32)*5/9
    print("Temperature in Celsius:",result)
else:
    print("Invalid Choice")
```

```python
#Q11. Area and Perimeter Calculator
# Take radius of circle from user and calculate:
# Area and Circumference of circle
# (Use pi = 3.14)

import math
radius = float(input("Enter the radius:"))
area = math.pi*radius**2
perimeter = 2*math.pi*radius
print("Area =",area)
print("Perimeter =",perimeter)
```

```python
#Q12:Student Grade Calculator
# Take marks of 5 subjects.
# Calculate percentage and print grade:
# >=90 → A+, >=80 → A, >=70 → B, >=60 → C, else → Fail
marks = float(input("Enter your marks: "))

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    print("Fail")
print("Grade=", grade)
```

```python
#Q13:Swap Two Variables
# Take two numbers from user and swap them using:
# a) Third variable
# b) Without third variable (Python way)
a = int(input("Enter the first number:"))
b = int(input("Enter the second number:"))

print("Before swapping:")
print("a=",a)
print("b=",b)
a,b=b,a
print("After swapping:")
print("a=",a)
print("b=",b)
```
