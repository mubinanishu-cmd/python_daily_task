```python
#21. (Hard) Create a dictionary student with 4 key-value pairs. Then use
#update() to add another dictionary, remove the last item using popitem(),
#and finally print all items using items().

dicts = {
    "name": "Nishhu",
    "age": 19,
    "course": "CSE",
    "city": "Mumbai"
}
dicts.update({"mail":123456})
print(dicts)
dicts.popitem()
print(dicts)
```

```python
#22. (Hard) Create a tuple with 8 numbers. Find the index of the number
#25. If it exists, count how many times it appears. Then convert the tuple
#into a list, sort it in descending order, and print.

tupl = (10, 25, 30, 40, 55, 25, 77, 88)
print(tupl.index(25))
print(tupl.count(25))
l = list(tupl)
l.sort(reverse=True)
print(l)
```

```python
#23. (Hard) Take 4 student names and their marks as input from the user.
#Store them in a dictionary (name as key, marks as value). Then find and
#print the student with the highest marks.

students = {}
for i in range(4):
    name = input("Enter the student name:")
    marks = int(input("Enter the student marks:"))
    students[name] = marks
print(students)

highest = max(students,key=students.get)

print("Student with highest marks:",highest)
print("Highest marks:",students[highest])
```

```python
#24. (Hard) Create a dictionary info = {"name": "Aman", "age": 21, "city":
#"Mumbai"}. Make a copy of it. Clear the original dictionary. Then add a
#new key "country": "India" in the copy and print both dictionaries.

dict_info =  {"name": "Aman", "age": 21, "city":"Mumbai"}
new_dict = dict_info.copy()
print(new_dict)
new_dict.update({"Country": "India"})
print(new_dict)
```

```python
#25. (Hard) Given data = (10, 20, 30, 20, 40, 20, 50), convert it into a
#dictionary where key is the number and value is its count of occurrences
#(using tuple methods + dictionary)

data = (10, 20, 30, 20, 40, 20, 50)
d = {}
for num in data:
    d[num] = data.count(num)
print(d)
```
