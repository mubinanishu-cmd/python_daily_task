```python
#1. Take a number as input. If the number is positive, check whether it is even or odd and print the result.

num = int(input("Enter a number: "))
if num > 0 :
    if num % 2 == 0:
        print("Positive Even")
    else:
        print("Position Odd")
else:
    print("Number is not postive")
```

```python
#2. Take age as input. If age is greater than or equal to 18, check if the person is a senior citizen (age >= 60) or adult.

age = int(input("Enter age: "))
if age >= 18:
    if age >= 60:
        print("Senior citizen")
    else:
        print("Adult")
else:
    print("Minor")
```

```python
#3. Take a single character. If it is an alphabet, check whether it is uppercase or lowercase.

ch = input("Enter a character: ")
if ch.isalpha():
    if ch.isupper():
        print("Uppercase")
    else:
        print("Lowercase")
else:
    print("Not an alphabet")
```

```python
#4. Take three numbers as input and print the largest number using nested if statements (without using max()).

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a >= b:
    if a >= c:
        print("Largest",a)
    else:
        print("Largest",c)
else:
    if b >= c:
        print("Largest",b)
    else:
        print("Largest",c)      
```

```python
#5. Take student marks. If marks >= 40 (Pass), then further check if marks >= 75 to print "Pass with Distinction".

marks = int(input("Enter student marks: "))
if marks >=40:
    if marks >=75:
        print("Pass Distinction")
    else:
        print("Pass")
else:
    print("Fail")
```

```python
#6. Take a number. If it is divisible by 2, check if it is also divisible by 4.

num = int(input("Enter a number: "))
if num % 2 == 0:
    if num % 4 == 0:
        print("Divisible by 2 and 4")
    else:
        print("Divisible by 2 but not by 4")
else:
    print("Not divisible by 2")
```

```python
#7. Take username and password. If username is "admin", then check if the password is correct.
username = input("Enter a username : ")
password = int(input("Enter a password: "))
if username == "admin":
    if password == "correct_password":
        print("Login Successful")
    else:
        print("Incorrect password")
else:
    print("Invalid username")
```

```python
#8. Take three sides of a triangle. First check if they form a valid triangle, then check if it is Equilateral, Isosceles, or Scalene.

a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a == b:
    if  b == c:
        print("Equilateral triangle")
    else:
        print("Isosceles triangle")
else:
     print("Scalene")
```

```python
#9. Take salary and years of experience. If salary > 50000, check experience to decide "Manager" or "Senior Manager".

salary = int(input("Enter a salary: "))
experience = int(input("Enter years of experience: "))
if salary >=50000:
    if experience == 5:
        print("Manager")
    else:
        print("Fresher")
else:
    print("Senior Manager")
```

```python
# 10.Take a character. If it is a letter, check if it is vowel or consonant. If it is a digit,
# print "Digit". Else print "Special Character".

ch = input("Enter a character: ")
if ch.isalpha():
    if ch.lower() in "aeiou":
        print("Vowel")
    else:
        print("Consonant")
elif ch.isdigit():
     print("Digit")
else:
    print("Special Character")
```
