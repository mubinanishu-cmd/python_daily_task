```python
#5. (Medium) Create two sets A = {1, 2, 3, 4} and B = {3, 4, 5, 6}. Print union, intersection, and difference.

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A.union(B))     #union
print(A.intersection(B))    #intersection
print(A.difference(B))    #difference
```

```python
#8.(Hard) Take 5 numbers as input from user, store them in a set, and print unique numbers only.

s = set()

for i in range(5):
    n = int(input("Enter a number: "))
    s.add(n)
print("Unique numbers:",s)
```

#**If Statement Questions (11-20):**

```python
#11. (Easy) Take age as input. If age >= 18, print "You are eligible to vote".

age = int(input("Enter the age: "))
if age >=18:
    print("You are eligible to vote")
else:
    print("Not eligible")
```

```python
#12. (Easy) Take a number and check if it is positive, negative, or zero using if-elif-else.

num = int(input("Enter a number: "))
if num > 0:
    print("Postive")
elif num < 0:
    
    print("Negative")
else:
    print("Zero")
```

```python
#13. (Medium) Take two numbers and print the larger one using if-else.

num1 = int(input("Enter the first number:"))
num2 = int(input("Enter the second number"))
if num1 > num2:
    print("Largest number is:",num1)
else:
    print("Largest number is:",num2)
```

```python
#14. (Medium) Take a character and check if it is a vowel (a, e, i, o, u) using if statement.

ch = input("Enter the character:")
if ch in "aeiou":
    print("Vowel")
else:
    print("Not a vowel")
```

```python
#15. (Medium) Take marks and print grade: ≥90 → A, ≥80 → B, ≥70 → C, else →Fail.

marks = int(input("Enter the marks: "))

if marks >=90:
    print("Grade A")
elif marks >=80:
    print("Grade B")
elif marks >=70 :
    print("Grade C")
else:
    print("Fail")
```

```python
#16. (Hard) Take three numbers and print the largest one using if-elif-else.

a = int(input("Enter the first number: "))
b = int(input("Enter the second number: "))
c = int(input("Enter the third number: "))

if a > b > c:
    print("Largest number is:",a)
elif b > a > c:
    print("Largest number is:",b)
else:
    print("Largest number is:",c)
```

```python
#17. (Hard) Take a number and check if it is even and greater than 50.

num = int(input("Enter the number:"))
if num % 2 == 0 and num > 50:
    print("The number is even and greater than 50")
else:
    print("Condition not satisfied")
```

```python
#18. (Hard) Take username and password. If username == "admin" and password =="1234", print "Login Successful", else "Login Failed".

username = input("Enter the username :")
password = input("Enter the password: ")

if username == "admin" and password == "1234":
    print("Login Successful ")
else:
    print("Login Failed")
```

```python
#19. (Hard) Take a year as input and check if it is a leap year (divisible by 4 but not 100, or divisible by 400).

year = int(input("Enter a year: "))

if year % 400 == 0:
    print("Leap Year")
elif year % 100 == 0:
    print("Not a Leap Year")
elif year % 4 == 0:
    print("Leap Year")
else:
    print("Not a leap year")
```

```python
#20. (Hard) Take three sides of a triangle and check if they can form a valid triangle (sum of any two sides > third side.

a = int(input("Enter first side: "))
b = int(input("Enter the second side: "))
c = int(input("Enter the third side: "))

if a + b > c and a + c > b and b + c > a:
    print("It is a triangle")
else:
    print("Not a triangle")
```
