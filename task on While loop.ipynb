4. Reverse a Number
Question
Write a program to reverse a number using a while loop.
Explanation
Example:
1234 → 4321
Steps:
• Take last digit using % 10
• Add digit into reverse number
• Remove last digit using // 10

```python
num = 1234
reverse = 0
while num > 0:
    reverse = reverse * 10 + num % 10
    num = num // 10
print(reverse)
```

```python
num = 1234
count = 0
while num > 0:
    count = count + 1
    num = num // 10
print(count)
```

```python
num = 121
original = num
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
if original == reverse:
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
balance = 1000

while True:
    print("1. Balance")
    print("2. Withdraw")
    print("3. Exit")

    choice = int(input("Choice: "))

    if choice == 1:
        print(balance)

    elif choice == 2:
        amount = int(input("Amount: "))
        balance = balance - amount
        print("Remaining:", balance)

    elif choice == 3:
        break
```
