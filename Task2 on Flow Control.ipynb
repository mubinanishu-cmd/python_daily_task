```python
#6. Write a program to display:
#o "Morning" if time < 12
#o "Afternoon" if time between 12 and 5
#o "Evening" otherwise

time = int(input("Enter time:"))
if time <= 12:
    print("Morning")
elif time >=12 and time <= 17:
    print("Afternoon")
else:
    print("Evening")
```

```python
#7. Write a program to check whether a student passed with distinction.
#o 75+ → Distinction
#o 35–74 → Pass
#o Below 35 → Fail

marks = int(input("Enter the marks: "))
if marks >= 75:
    print("Distinction")
elif marks >=35 and marks <= 75:
    print("Pass")
else:
    print("Fail")                    
```

```python
#1. Write a program to find the largest among three numbers.
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
num3 = int(input("Enter third number: "))

if num1 > num2 > num3:
    print("Largest number: ",num1)
elif num2 > num3 > num1:
    print("Largest number: ",num2)
elif num3 > num2 > num1:
    print("Largest number:",num3)
else:
    print("Smallest number")
```

```python
#2. Write a program to calculate electricity bill:
# o Units < 100 → ₹2/unit
# o 100–300 → ₹4/unit
# o Above 300 → ₹6/unit

unit = int(input("Enter a Number: "))
if unit <= 100:
    print("₹2/unit")
elif unit >=100 and unit <=300:
    print("₹4/unit")
else:
    print("₹6/unit")
```

```python
#3. Write a program to calculate bonus:
# o Salary > 50000 → 20% bonus
# o Salary between 30000–50000 → 10% bonus
# o Otherwise → 5% bonus

salary = int(input("Enter a Salary : "))
if salary >= 50000:
    print("20% bonus")
elif salary >=30000 and salary >=50000:
    print("10% bonus")
else:
    print("5% bonus") 
```

```python
#5. Write a program to check whether a number is:
# o Multiple of both 3 and 5
# o Multiple of only 3
# o Multiple of only 5
# o None

wh = int(input("Enter the Whether: "))
if wh % 3 ==0 and wh % 5 == 0:
    print("Multiple of both")
elif wh % 3 ==0:
    print("Multiple of only 3")
elif wh % 5 ==0:
    print("Multiple of only 5")
else:
    print("None")
```

```python
#6. Write a program to find whether a person is:
# o Child (0–12)
# o Teenager (13–19)
# o Adult (20–59)
# o Senior Citizen (60+)

age = int(input("Enter a person age: "))
if age >=0 and age <= 12:
    print("Child")
elif age >=13 and age <=19:
    print("Teenager")
elif age <=60:
    print("Senior Citizen")
else:
    print("Old")
```

```python
#7. Write a program to calculate ticket price:
# o Age below 5 → Free
# o 5–18 → ₹100
# o Adults → ₹200
# o Senior citizens → ₹150

age = int(input("Enter an age: "))
if age <=5:
    print("Free")
elif age >=5 and age <=18:
    print("Price: ₹100")
elif age <= 19:
    print("Price: ₹200")
elif age <= 60:
    print("Price : ₹150 ")
else:
    print("Not allow")
```
