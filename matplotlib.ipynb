```python
import matplotlib.pyplot as plt

#pyplot is a collection of functions that makes matplotlib work as MATLAB
```

#**Create first Chart:**

```python
#data
x=[1,2,3,4,5,6]
y=[10,20,30,45,55,60]

plt.plot(x,y)  #Line chart
plt.show()    #display chart 
```

#**Customize Chart:**

```python
plt.figure(figsize=(6,2))   #figure size
plt.plot(x, y, color='orange', marker='o', linestyle='-',linewidth=2, markersize=10)#coloring chart

plt.title("GRAPH")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#data2
x=[1,2,3,4,5,6]
y1=[10,20,30,45,55,60]
y2=[10,25,30,45,55,65]

plt.plot(x,y1,label='Sales2025')
plt.plot(x,y1,label='Sales2024')

plt.title("BSales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.legend()
plt.show()
```
