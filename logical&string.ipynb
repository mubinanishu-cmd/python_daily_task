#**Logical Operators (and, or, not) Questions:**

```python
#Question 1:
# Take age and marks from user.
# Check if the student is eligible for admission:
# Age must be between 18 and 25 AND marks >= 60
age = int(input("Enter the age: "))
marks = int(input("Enter the marks :"))

if age >=18 and age<=25 and marks >=60:
    print("Student is eligible for admission")
else:
    print("Student is not eligible for admission")
```

```python
#Question 2:
# Take username and password from user.
# Check login:
# Username == "admin" AND Password == "12345"
# Print "Login Successful" or "Login Failed"
username = input("Enter the username: ")
password = int(input("Enter the password: "))

if username=="admin" and password == "12345":
    print("Login Successful")
else:
    print("Login Failed")
```

```python
# Question 3:
# Take a number from user.
# Check if the number is between 50 and 100 (using 'and')
# Also check if it is NOT divisible by 5 (using 'not')
num = int(input("Enter the number: "))
if num >=50 and num<=100 and not (num%5==0):
    print("Divisible by 5")
else:
    print("Not Divisible by 5")
```

```python
#Question 4:
# Take three numbers from user.
# Print the largest number using 'and' and 'or' operators.
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
num3 = int(input("Enter the third number: "))
if (num1>=num2) and (num1>=num3):
    print("Largest Number:",num1)
elif (num2>=num1) and (num2>=num3):
    print("Largest Number:",num2)
else:
    print("Largest Number:",num3)
    
```

#2. String and String Slicing Questions*

```python
#Question 5:
# Take your full name as input from user.
# Print the following:
# 1. Full name in uppercase
# 2. Full name in lowercase
# 3. Length of your name
# 4. First 3 characters
# 5. Last 3 characters

name = input("Enter the Full_name:")
print("Uppercase:",name.upper())
print("Lowercase:",name.lower())
print("Length:",len(name))
print("First 3 character:",name[:3])
print("Last 3 character:",name[-3:])
```

```python
#Question 6:
# Take any sentence from user.
# Print:
# - First 10 characters
# - Last 10 characters
# - Characters from index 5 to 15
# - Every 2nd character of the sentence
user = input("Enter the sentence:")
print("First 10 letters:",user[:10])
print("Last 10 letters:",user[-10:])
print("Characters from the index:",user[5:16])
print("Every 2nd character of the sentence:",user[::2])
```

```python
#Question 7:
# Take a word from user.
# Check if the word is a palindrome or not.
# (Example: "madam", "radar")
word = input("Enter the Word:")
if word ==word[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
#Question 8:
# Take email as input (e.g: student@gmail.com)
# Print only the username (part before @)
# Print only the domain (part after @)
email = "mubina@gmail.com"
print("Only the username:",email[:6])
print("Only the domain:",email[6:])
```

```python
#Question 9:
# Take a password from user.
# Check if password is strong:
# - Length >= 8
# - Contains '@' or '#'
# Print "Strong Password" or "Weak Password"
password = input("Enter the password:")
if len(password )>=8 and ("@" in password or "#" in password):
    print("Strong Password")
else:
    print("Weak Password")
```

```python
#Q1. Logical Operators
# Take three subject marks from user (English, Math, Science).
# Check if the student has passed:
# All marks should be >= 40 (use 'and' operator)
# If any one subject is < 40, print "Fail"
english = int(input("Enter the marks: "))
math = int(input("Enter the marks: "))
science = int(input("Enter the marks:"))

if english >=40 and math >=40 and science >=40:
    print("Pass")
else:
    print("Fail")
```

```python
#Q2. Logical Operators + User Input
# Take a number from user.
# Check whether the number is:
# Positive AND Even
# OR
# Negative OR Odd
# Print appropriate message

num = int(input("Enter the number: "))
if num >0 and num % 2 == 0:
    print("Postive and Even")
elif num > 0 and num % 2 !=0:
    print("Postive and Odd")
elif num <0 and num % 2 == 0:
    print("Negative and Even")
elif num<0 and num % 2 !=0:
    print("Negative and Odd")
else:
    print("Zero")
```

```python
#Q3. String Operations
# Take your city name as input.
# Print the following:
# 1. City name in reverse order
# 2. Number of vowels in the city name (a,e,i,o,u)
# 3. City name without first and last character

city = "Hyderabad"
count = 0
print("Reverse_Order:",city[::-1])

for ch in city.lower():
    if ch in "aeiou":
        count +=1
print("Number of vowels:",count)

print("Without first charater:",city[0])
print("Without first charater:",city[-1])
```

```python
#Q4. String Slicing
# Take a mobile number as input (10 digits).
# Print:
# First 3 digits (network code)
# Last 4 digits
# Middle 3 digits
number = input("Enter the Mobile_number: ")
print("First 3 digits:",number[:3])
print("Last 4 digits:",number[-4:])
print("Middle 3 digits:",number[3:6])
```

```python
#Q5. Mixed (String + Logical)
# Take a string from user.
# Check if:
# - The string length is greater than 5
# - AND it starts with 'A' or 'a'
# Print "Valid String" or "Invalid String"

word = input("Enter the String:")
if len(word)>5 and (word[0]) == 'A' or word[0] =='a':
    print("Valid String")
else:
    print("Invalid String")
```
