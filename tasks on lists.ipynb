```python
#11. Sort the list [45, 12, 78, 23, 56] in ascending order using sort().

list = [45, 12, 78, 23, 56]
list.sort()
print(list)
```

```python
#12. Sort the list [45, 12, 78, 23, 56] in descending order.

list = [45, 12, 78, 23, 56]
list.sort(reverse=True)
print(list)
```

#**Hard / Mixed Level (16-20):**

```python
#16. Create a list of 5 numbers. Append 3 more numbers, then sort it in descending order.

list = [1, 2, 3, 4, 5]
list.extend([6, 7, 8])
list.sort(reverse=True)
print(list)
```

```python
#17. Create a list. Insert "Python" at index 0, then remove the last element
#using pop(), and print the length.

list = ["Java", "C++", "Ruby", "C#", "DataScience"]
list.insert(0, "python")
print(list)

print(list.pop())

print(len(list))
```

```python
#18. Take 5 city names as input from user and store them in a list. Then print them in reverse order.

cities = []
city1 = input("Enter the 1 city names: ")
city2 = input("Enter the 2 city names: ")
city3 = input("Enter the 3 city names: ")
city4 = input("Enter the 4 city names: ")
city5 = input("Enter the 5 city names: ")

cities.append(city1)
cities.append(city2)
cities.append(city3)
cities.append(city4)
cities.append(city5)

cities.reverse()
print(cities)
```

```python
#19. Given data = [10, 20, 30, 20, 40, 20], remove all occurrences of 20 (use loop or multiple remove).

data =  [10, 20, 30, 20, 40, 20]
data.remove(20)
while 20 in data:
    data.remove(20)
print(data)
```

```python
#20. Create a list of student marks. Sort it, find the highest and lowest
#mark using list methods, and print them.

marks = [75, 85, 90, 65, 92]
marks.sort()
print("Sorted marks:",marks)
print("Highest marks:",marks[-1])
print("Lowest marks:",marks[0])
```
