#**String Slicing Questions:**

```python
#1. Take a string from user and print the first 5 characters.
#2. Take a string and print the last 4 characters using negative slicing.
#3. Take a string and print characters from index 2 to 8 (excluding 8).
#4. Take a string and print every 2nd character of the string.
#5. Take a full name (e.g. "Sumit Kumar Sharma") and print only the first
#name and last name separately using slicing.

#1.
name = "MubinaBegum"
print(name[:5])
#2.
name = "MubeenaNishu"
print(name[-4:])
#3.
print(name[2:8])
#4.
print(name[::2])
print(name[::1])
#5.
name = "Mubina Nishu Mahi"
print(name[:6])
print(name[:13])
```

#**Case Conversion Questions:*

```python
#6. Take a string from user and print it in:
#• Uppercase
#• Lowercase
#• Title Case
#• Capitalized (only first letter capital)

name = "mubeena begum"
print(name.upper())
print(name.lower())
print(name.title())
print(name.capitalize())
```

```python
#11. Take a string and check if it is alphanumeric (isalnum()).
m = "Java6798"
print(m.isalnum())
```

```python
#12. Take a string and check if it is completely in lowercase (islower()).
n = "mubeenaBegum"
print(n.islower())
```

```python
#13. Take a password as input and check:
#• It should be alphanumeric
#• It should be minimum 8 characters long Print "Valid Password" or "Invalid Password".

password = input()

if password.isalnum() and len(password) >= 8:
    print("Valid Password")
else:
    print("Invalid Password")
```

```python
#20. Take any string from user.
#• Print the string in reverse order using slicing
#• Convert it to title case
#• Check if the reversed string is same as original (Palindrome check)

text = "Hello python"
print(text[::-1])

print(text.title())

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```
