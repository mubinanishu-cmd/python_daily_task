#**Here are 20 Questions on for Loop and while Loop in Python**

```python
# 1. Write a for loop to print numbers from 1 to 10.

for i in range(11):
    print(i)
```

```python
# 2. Use for loop to print all even numbers between 1 to 20.

for i in range(21):
    print(i)
```

```python
# 3. Take a number from user and print its multiplication table using for loop (up to 10).

num = 10
i = 1
while i <= 10:
    print(num * i)
    i = i + 1   
```

```python
# 4. Use for loop to calculate the sum of first 10 natural numbers.

i = 2 
total = 0
while i <= 10:
    total = total + i
    i = i + 2
print(total)
```

```python
# 5. Print the following pattern using for loop:
i = 1
while i <= 5:
    print("*"* i)
    i = i + 1
```

```python
# 6. Take a list of 5 numbers and print their squares using for loop.
num = [1, 2, 3, 4, 5]
for n in num:
    print(n ** 2)
```

```python
# 7. Use for loop with range() to print numbers from 50 down to 30 (decreasing).

for i in range(50, 29, -1):
    print(i)
```

```python
#8. Count and print how many numbers between 1 to 50 are divisible by 3 using for loop.

count = 0
for i in range(1,51):
    if i % 3 == 0:
        count +=1
print("Count:",count)
```

```python
#9. Take a string from user and print each character using for loop.

s = input("Enter a string: ")
for ch in s:
    print(ch)
```

```python
#10. Use nested for loop to print multiplication table from 1 to 5.

for i in range(1, 6):
    for j in range(1, 11):
        print(f"{i} x {j} = {i * j }")
    print()
```

#**While Loop Questions (11-20)**

```python
# 11. Write a while loop to print numbers from 1 to 15.
i = 1
while i <= 15:
    print(i)
    i += 1
```

```python
# 12. Take a number from user and print its multiplication table using while loop.
num = int(input("Enter a number: "))
i = 1
while i <= 10:
    print(f"{num} x {i} = {num*i}")
    i +=1
```

```python
# 13. Use while loop to find the sum of numbers until user enters 0.
total = 0
while True:
    n = int(input("Enter a number (0 to stop): "))
    if n == 0:
        break
    total += n
print("Sum:",total)
```

```python
# 14. Print all odd numbers from 1 to 20 using while loop.
i = 1
while i <= 20:
    if i % 2 != 0:
        print(i)
    i += 1
```

```python
#15. Take a positive number and keep dividing it by 2 using while loop until it becomes less than 1.Count the steps.

n = float(input("Enter a positive number: "))
steps = 0
while n >= 1:
    n /= 2
    steps += 1
print("Steps:", steps)
```

```python
## 16. Use while loop to reverse a number (e.g., 1234 → 4321).

n = int(input("Enter a number: "))
reversed_num = 0
temp = n
while temp > 0:
    digit = temp % 10
    reversed_num = reversed_num * 10 + digit
    temp //= 10
print("Reversed:", reversed_num)
```

```python
#17. Keep taking input from user until they enter "stop" (using while loop).

while True:
    text = input("Enter text (or 'stop' to end): ")
    if text.lower() == "stop":
        break
    print("You entered:", text)
```

```python
#18. Print the Fibonacci series up to 10 terms using while loop.

a, b = 0, 1
count = 0
while count < 10:
    print(a, end=" ")
    a, b = b, a + b
    count += 1
```

```python
#19. Use while loop to check if a given number is prime or not.

n = int(input("Enter a number: "))
if n <= 1:
    print(n, "is not prime")
else:
    i = 2
    is_prime = True
    while i * i <= n:
        if n % i == 0:
            is_prime = False
            break
        i += 1
    if is_prime:
        print(n, "is prime")
    else:
        print(n, "is not prime")
```

```python
#20. Create a simple menu-driven program using while loop (options: 1. Add, 2. Subtract, 3.Exit).

while True:
    print("\n1. Add")
    print("2. Subtract")
    print("3. Exit")
    choice = int(input("Enter your choice: "))

    if choice == 1:
        a = float(input("Enter first number: "))
        b = float(input("Enter second number: "))
        print("Result:", a + b)
    elif choice == 2:
        a = float(input("Enter first number: "))
        b = float(input("Enter second number: "))
        print("Result:", a - b)
    elif choice == 3:
        print("Exiting program...")
        break
    else:
        print("Invalid choice, try again.")
```
